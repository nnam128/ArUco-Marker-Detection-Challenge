```python
import os
import pandas as pd
from concurrent.futures import ThreadPoolExecutor, as_completed

from src.pipeline import run_aruco_pipeline
from src.postprocess import load_cnn_model


# CONFIG

MODEL_PATH = "models/aruco_classifier_model_epoch50.keras"
TRAIN_DIR = "data/aruco_data/test"

OUTPUT_CSV = "submission.csv"

NUM_IMAGES = 500
MAX_WORKERS = 15


# LOAD MODEL ONCE

cnn_model = load_cnn_model(MODEL_PATH)


# SINGLE IMAGE PROCESS

def process_single_image(img_path):
    img_name = os.path.basename(img_path)
    image_id = os.path.splitext(img_name)[0]

    try:
        result = run_aruco_pipeline(
            img_path,
            cnn_model
        )

        if not result:
            result = " "

        print(f"[OK] {img_name}")

    except Exception as e:
        print(f"[ERROR] {img_name}: {e}")
        result = " "

    return {
        "image_id": image_id,
        "prediction_string": result
    }


# PARALLEL INFERENCE

def run_parallel_batch_inference(
    image_dir,
    output_csv=OUTPUT_CSV,
    num_images=NUM_IMAGES,
    max_workers=MAX_WORKERS
):
    image_files = sorted([
        f for f in os.listdir(image_dir)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ])[:num_images]

    image_paths = [
        os.path.join(image_dir, f)
        for f in image_files
    ]

    print(f"Total images: {len(image_paths)}")
    print(f"Workers: {max_workers}\n")

    results = []

    with ThreadPoolExecutor(max_workers=max_workers) as executor:
        futures = [
            executor.submit(process_single_image, img_path)
            for img_path in image_paths
        ]

        for future in as_completed(futures):
            results.append(future.result())

    results = sorted(results, key=lambda x: x["image_id"])

    df = pd.DataFrame(results)
    df.to_csv(output_csv, index=False)

    print("\n================================")
    print(f"Saved to: {output_csv}")
    print("================================")

    return df


# MAIN

df_result = run_parallel_batch_inference(
    image_dir=TRAIN_DIR,
    output_csv=OUTPUT_CSV,
    num_images=NUM_IMAGES,
    max_workers=MAX_WORKERS
)

print(df_result.head())
```
